```python
import pandas as pd

ruta_archivo = '../data/ventas_brutas.csv'
df = pd.read_csv(ruta_archivo)

df.head()
```

```python
nulos_por_columna = df.isnull().sum()
print("Valores faltantes por columna:")
print(nulos_por_columna)
```

```python
df = df.dropna(how='any')

print(f"filas resultantes: {len(df)}")
```

```python
df['Fecha_Compra'] = pd.to_datetime(df['Fecha_Compra'])

df['Mes'] = df['Fecha_Compra'].dt.month

df['Venta_Total'] = df['Cantidad'] * df['Precio_Unitario']

df.head()
```

```python
df.to_csv('../data/ventas_limpias.csv', index=False)
print("Archivo 'Ventas_Limpias' creado con exito!")
```
