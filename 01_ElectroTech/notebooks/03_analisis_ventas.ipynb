```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv('../data/ventas_limpias.csv')

sns.set_theme(style="whitegrid")
```

```python
ventas_por_mes = df.groupby('Mes')['Venta_Total'].sum()

plt.figure(figsize=(10, 5))
ventas_por_mes.plot(kind='bar', color='skyblue')

plt.title('Ventas Totales por Mes', fontsize=15)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Ventas en USD ($)', fontsize=12)
plt.xticks(rotation=0)
plt.show()
```

```python
ventas_ciudad = df.groupby('Direccion')['Venta_Total'].sum().reset_index()

ventas_ciudad = ventas_ciudad.sort_values(by='Venta_Total', ascending=False)

plt.figure(figsize=(10, 5))
sns.barplot(data=ventas_ciudad, x='Direccion', y='Venta_Total', palette='viridis')

plt.title('Ingresos Totales por Ciudad', fontsize=15)
plt.show()
```

```python
ventas_por_mes.plot(kind='bar', color='skyblue', figsize=(10, 5))
plt.title('Ventas Totales por Mes')
plt.savefig('../reports/Ventas_mensuales.png')
plt.close()

plt.figure(figsize=(10, 5))
sns.barplot(data=ventas_ciudad, x='Direccion', y='Venta_Total', palette='viridis')
plt.title('Ingresos por Ciudad')
plt.savefig('../reports/ventas_por_ciudad.png')
plt.show()
```
