```python
import pandas as pd
import numpy as np
import os
from datetime import datetime, timedelta

# Configuración de nombres y ciudades
productos = {
    'iPhone 15': 999,
    'MacBook Pro': 1499,
    'Google Pixel 8': 699,
    'AirPods Pro': 249,
    'Monitor 4K': 399,
    'Teclado Mecánico': 120,
    'Cargador USB-C': 25
}

ciudades = ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Bilbao']

# Generación de datos aleatorios
n_filas = 10000
data = {
    'ID_Orden': np.arange(1000, 1000 + n_filas),
    'Producto': np.random.choice(list(productos.keys()), n_filas),
    'Cantidad': np.random.randint(1, 5, n_filas),
    'Fecha_Compra': [datetime(2025, 1, 1) + timedelta(minutes=np.random.randint(0, 525600)) for _ in range(n_filas)],
    'Direccion': np.random.choice(ciudades, n_filas)
}

df = pd.DataFrame(data)

# Añadir columna de Precio Unitario basado en el diccionario
df['Precio_Unitario'] = df['Producto'].map(productos)

# --- ENSUCIAR LOS DATOS (Para practicar limpieza después) ---
# Vamos a meter 500 filas con valores nulos (NaN)
for _ in range(500):
    idx = np.random.randint(0, n_filas)
    df.loc[idx, 'Producto'] = np.nan

# Guardar en la carpeta data
output_path = '../data/ventas_brutas.csv'
df.to_csv(output_path, index=False)

print(f"✅ Archivo creado exitosamente en: {output_path}")
print(f"Registros generados: {len(df)}")
```
